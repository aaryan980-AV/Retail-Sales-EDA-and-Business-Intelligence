```python
import pandas as pd

df = pd.read_csv("cleaned_sales_data (1).csv")

print("Dataset Loaded Successfully!")
print(df.head())
print("Dataset Shape:", df.shape)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# Correlation Heatmap
plt.figure(figsize=(12, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Sales Data Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Age',
    y='Total_Sales',
    hue='Gender',
    s=80
)

plt.title("Customer Age vs Total Sales")
plt.xlabel("Customer Age")
plt.ylabel("Total Sales")
plt.legend(title="Gender")
plt.grid(alpha=0.3)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
selected_columns = [
    'Age',
    'Quantity',
    'Unit_Price',
    'Total_Sales'
]

# Create Pair Plot
sns.pairplot(
    df[selected_columns],
    diag_kind='hist',
    corner=True
)

plt.suptitle("Pair Plot - Sales Dataset", y=1.02)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Category-wise Total Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
city_sales = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_sales.index,
    y=city_sales.values
)

plt.title("City-wise Total Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```
